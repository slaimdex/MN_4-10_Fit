## 11. Підсумкові завдання

### Завдання 36
Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
# Код студента
import numpy as np


sales = np.array([
    [120, 150, 170, 160],
    [80,  90,  100, 110],
    [200, 180, 210, 230]
])

items = ['Товар 1', 'Товар 2', 'Товар 3']
times = ['Місяць 1', 'Місяць 2', 'Місяць 3','Місяць 4' ]

tovar1=sales[0].sum()
tovar2=sales[1].sum()
tovar3=sales[2].sum()

Mounth1=sales[:, 0].sum()
Mounth2=sales[:, 1].sum()
Mounth3=sales[:, 2].sum()
Mounth4=sales[:, 3].sum()


values = [tovar1, tovar2, tovar3]

mounth = [Mounth1,Mounth2,Mounth3,Mounth4]

best_index=np.argmax(values)
best_mounth=np.argmax(mounth)


print("\n","Сума товару 1:", tovar1)
print("Сума товару 2:", tovar2)
print("Сума товару 3:", tovar3)

print("\n","Сума продажів першого місяця:", sales[:, 0].sum())
print("Сума продажів другого місяця:", sales[:, 1].sum())
print("Сума продажів третього місяця:", sales[:, 2].sum())
print("Сума продажів четвертого місяця:", sales[:, 3].sum())


print("\n",f"Найкраще продається {items[best_index]} із сумою {sales[best_index].sum()}")
print(f"Найкраще продається {times[best_mounth]} із сумою {sales[:, [best_mounth]].sum()}")

print("\n","Середня сума продажу товару1:",sales[0].mean())
print("Середня сума продажу товару2:",sales[1].mean())
print("Середня сума продажу товару3:",sales[2].mean())
````

Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента

rng = np.random.default_rng(seed=42)

integers = rng.integers(1, 20, size=25)

m =  integers.reshape(5, 5)


print("\n",m)
print("\n","Сума діагоналі:", np.diag(m).sum())

flip_m = np.fliplr(m)

print("\n","Побічна діагональ:", np.diag(flip_m).sum())

print("\n","найбільший аргумент першого рядка:", m[0].max())
print("найбільший аргумент другого рядка:", m[1].max())
print("найбільший аргумент третього рядка:", m[2].max())
print("найбільший аргумент четвертого рядка:", m[3].max())
print("найбільший аргумент п'ятого рядка:", m[4].max())

print("\n","найменьший аргумент першого стовбця:", m[:, 0].min())
print("найменьший аргумент другого стовбця:", m[:, 1].min())
print("найменьший аргумент третього стовбця:", m[:, 2].min())
print("найменьший аргумент четвертого стовбця:", m[:, 3].min())
print("найменьший аргумент п'ятого стовбця:", m[:, 4].min())

a = np.mean(m)
view = m[:]
view[m<a] = 0

print ("\n","Середнє значення в матриці:" , a)
print ("\n",m)
````

### Завдання 39

```text

3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
