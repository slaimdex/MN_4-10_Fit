### Завдання 25
Дано два вектори:

```text

```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
# Код студента
import numpy as np

a = np.array([1, 2])
b = np.array([2, 1])

dot = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

cos = dot / (norm_a * norm_b)

angle = np.degrees(np.arccos(cos))

print("Кут у градусах:", angle)
print("Косинус кута:", cos)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
import numpy as np

a = np.array( [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]])

mu = a.mean(axis=1)
sigma = a.std(axis=1)
a_std = (a - mu) / sigma

print("Середні після стандартизації:")
print(a_std.mean(axis=1))

print("\nСтандартні відхилення:")
print(a_std.std(axis=1))
````
